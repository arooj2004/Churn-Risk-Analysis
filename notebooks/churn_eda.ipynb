```python
# -------------------------------------------------------------
# 1. Feature Engineering: Assigning Points based on Data Shapes
# -------------------------------------------------------------

# Tenure Score (Tiered cliff)
df["Score_Tenure"] = np.select(
    [df["tenure"] <= 12, (df["tenure"] > 12) & (df["tenure"] <= 24)],
    choicelist=[2, 1],
    default=0,
)

# Monthly Charges Score (Single accelerated threshold)
df["Score_Charges"] = np.where(df["MonthlyCharges"] >= 60, 1, 0)

# Contract Score (Highest risk for Month-to-Month)
df["Score_Contract"] = np.select(
    [df["Contract"] == "Month-to-month", df["Contract"] == "One year"],
    choicelist=[2, 1],
    default=0,
)

# Internet Service Score (Fiber optic friction)
df["Score_Internet"] = np.where(df["InternetService"] == "Fiber optic", 1, 0)

# Payment Method Score (Electronic check friction)
df["Score_Payment"] = np.where(df["PaymentMethod"] == "Electronic check", 1, 0)

# Senior Citizen Score
df["Score_Senior"] = np.where(df["SeniorCitizen"] == 1, 1, 0)

# -------------------------------------------------------------
# 2. Aggregate into Final Risk Score (Max possible score = 8)
# -------------------------------------------------------------
score_columns = [
    "Score_Tenure",
    "Score_Charges",
    "Score_Contract",
    "Score_Internet",
    "Score_Payment",
    "Score_Senior",
]
df["RiskScore"] = df[score_columns].sum(axis=1)

# -------------------------------------------------------------
# 3. Validation: Does the score cleanly correlate with Churn?
# -------------------------------------------------------------
# Assumes 'Churn' column is binary (1 for Churn, 0 for Retained)
if df["Churn"].dtype == "object":
    df["Churn"] = np.where(df["Churn"].str.lower() == "yes", 1, 0)

validation_table = (
    df.groupby("RiskScore")
    .agg(Total_Customers=("Churn", "count"), Churned_Customers=("Churn", "sum"))
    .reset_index()
)

# Calculate real-world churn rate per risk tier
validation_table["Actual_Churn_Rate"] = (
    validation_table["Churned_Customers"] / validation_table["Total_Customers"]
) * 100

# Display the validation matrix
print(validation_table.to_string(index=False))
```

```python
# -------------------------------------------------------------
# 4. Create Contact List (Active Customers Only)
# -------------------------------------------------------------

# Keep only customers who have NOT churned
active_customers = df[df["Churn"] == 0].copy()

# -------------------------------------------------------------
# 5. Assign Risk Tiers
# -------------------------------------------------------------
active_customers["RiskTier"] = np.select(
    [
        active_customers["RiskScore"] >= 6,
        active_customers["RiskScore"].between(4, 5),
        active_customers["RiskScore"].between(0, 3),
    ],
    [
        "call this week",
        "call this month",
        "monitor",
    ]
)

# -------------------------------------------------------------
# 6. Sort Highest-Risk Customers First
# -------------------------------------------------------------
active_customers = active_customers.sort_values(
    by=["RiskScore", "tenure"],
    ascending=[False, True]
)

# Preview contact list
print(
    active_customers[
        [
            "customerID",
            "RiskScore",
            "RiskTier",
            "tenure",
            "MonthlyCharges",
            "Contract",
        ]
    ].head(10)
)
```

```python
# -------------------------------------------------------------
# 7. Count Customers in Each Tier
# -------------------------------------------------------------

tier_counts = (
    active_customers["RiskTier"]
    .value_counts()
    .reindex(
        ["call this week", "call this month", "monitor"],
        fill_value=0
    )
    .reset_index()
)

tier_counts.columns = ["RiskTier", "Customers"]

print(tier_counts.to_string(index=False))
```
